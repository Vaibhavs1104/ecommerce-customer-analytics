```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv(r'C:\Sales Analysis\data\raw\Sales Transaction v.4a.csv')
print(df.shape)
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
print(df['Date'].min(), '→', df['Date'].max())
```

```python
missing_cust = df[df['CustomerNo'].isna()]
print(f"Rows with missing CustomerNo: {len(missing_cust)}")

df = df.dropna(subset=['CustomerNo']).copy()
df['CustomerNo'] = df['CustomerNo'].astype(int)
print(f"Remaining rows: {len(df)}")
```

```python
df['CustomerNo'].isna().sum()
```

```python
df.info()
```

```python
df['IsCancelled'] = df['TransactionNo'].astype(str).str.startswith('C')

df_cancelled = df[df['IsCancelled']].copy()
df_sales = df[~df['IsCancelled']].copy()

print(f"Cancelled line items: {len(df_cancelled)}")
print(f"Clean sales line items: {len(df_sales)}")
```

```python
print("Negative/zero quantity:", (df_sales['Quantity'] <= 0).sum())
print("Negative/zero price:", (df_sales['Price'] <= 0).sum())
```

```python
df_sales['Revenue'] = df_sales['Price'] * df_sales['Quantity']
df_sales[['ProductName', 'Price', 'Quantity', 'Revenue']].head()
```

```python
df_sales.to_csv(r'C:\Sales Analysis\data\processed\cleaned_sales.csv', index=False)
df_cancelled.to_csv(r'C:\Sales Analysis\data\processed\cancellations.csv', index=False)

print("Saved cleaned_sales.csv and cancellations.csv to data/processed/")
```
